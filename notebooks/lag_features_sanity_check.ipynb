```python
import pandas as pd

df = pd.read_parquet('../data/processed/feature_table_15min.parquet')

print(df.shape)
print(df.info())
```

```python
# Confirm the time series is actually perfectly regular (no gaps).
# Since we built this with a fixed step (arange), every gap between consecutive rows should be exactly 900 seconds (15 min). 
# If this print shows more than one number, something's wrong with the data.
diffs = df['bucket_start'].diff().dropna()
print(diffs.unique())
```

```python
# Autocorrelation of gpu_milli at increasing lags.
# Each lag = one 15-min bucket back, so lag=4 is 1 hour back,
# lag=8 is 2 hours back, lag=96 is roughly 1 full day back.
for lag in [1, 2, 4, 8, 12, 16, 24, 48, 96]:
    print(lag, df['gpu_milli'].autocorr(lag=lag))
```
